# 01 Data Preprocessing

This notebook reads the Amazon Reviews 2023 files for the three fixed categories, preprocesses them into parquet, and prints the exact statistics needed for the P1 data phase.

## Data copy section

Place the raw dataset files in:

- `data/raw/All_Beauty/All_Beauty.jsonl.gz`
- `data/raw/All_Beauty/meta_All_Beauty.jsonl.gz`
- `data/raw/Health_and_Personal_Care/Health_and_Personal_Care.jsonl.gz`
- `data/raw/Health_and_Personal_Care/meta_Health_and_Personal_Care.jsonl.gz`
- `data/raw/Handmade_Products/Handmade_Products.jsonl.gz`
- `data/raw/Handmade_Products/meta_Handmade_Products.jsonl.gz`

The pipeline also accepts the normalized `raw_review_*` and `raw_meta_*` filenames, but the original Amazon filenames work directly and are already supported.

## Imports and configuration

The notebook uses the shared `src/recsys` pipeline rather than defining any model or data logic inline.

```python
from pathlib import Path

from recsys.config import get_settings
from recsys.data.preprocess import preprocess_dataset

settings = get_settings()
settings.ensure_directories()

categories = ("All_Beauty", "Health_and_Personal_Care", "Handmade_Products")
print("Raw data:", settings.raw_data_dir)
print("Processed data:", settings.processed_data_dir)
print("Categories:", categories)
```

## Run preprocessing

This cell writes the parquet outputs into `data/processed/`.

```python
result = preprocess_dataset(
    settings.raw_data_dir,
    settings.processed_data_dir,
    categories=categories,
    max_interactions=1_000_000,
)

print(result.stats)
print("interactions:", result.interactions_path)
print("items:", result.items_path)
print("stats:", result.stats_path)
```

## Exported artifacts

The notebook writes the processed files into `data/processed/`:

- `interactions.parquet`
- `items.parquet`
- `stats.json`
